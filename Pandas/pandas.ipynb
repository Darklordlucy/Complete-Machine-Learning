## Add / Remove columns

```python
sample1 = sample.copy()
sample1['Address'] = "Usa"          # ---> Add same element to all values
print(sample1)
```

```python
import numpy as np
sample1['Company'] = np.where(sample1['Age'] > 30 , 'Microsoft' , 'Apple')      # ----> To add new col and data according to the condition
print(sample1)
```

```python
sample1 = sample.drop(columns=['Company','Age','ID'])
print(sample1)
```

```python
sample1 = sample1[['Name','Salary','Address']]
print(sample1)
```

```python
sample1 = sample.copy()
sample1['New_salary'] = sample1['ID'] /  sample1['Salary']
print(sample1)
```

```python
sample1['New_col'] = sample1['Name'].str.split(' ').str[0]
print(sample1.head())
```

```python
sample1.to_csv("path",index=False)
```

```python
sample1['new_col3'] = sample1['Salary'].apply(lambda X: 'Yes' if X  > 65000 else 'No')
sample1
```

```python
sample1['New'] = sample1['Name'].str.split(" ").str[0]
sample1.drop(columns='New')
```

```python
def fun(row):
    if row['Age'] > 30 and row['Salary'] > 65000:
        return 'Eligible'
    else:
        return 'Not Eligible'

sample1['New_col'] = sample1.apply(fun, axis=1)   # axis = 1 (rows) and axis = 0 (column)
sample1
```

## Aggregating Data

```python
sample = pd.read_csv("D:\\Machine Learning\\Pandas\\sample_file.csv")
sample = sample.copy() 
sample
```

```python
sample[sample['Category']=='Starter']['Item'].value_counts()
```

```python
sample.groupby(['Category'])['BillAmount'].sum()
```

```python
sample.groupby(['Category']).agg({'Quantity': 'sum', 'BillAmount': 'mean'})
```

## Advanced Funtionality

```python
sample['New_col'] = sample['Quantity'].shift(1)  # --- Shift exist col down in new col
```

```python
sample['New_col1'] = sample['Quantity'].rank(ascending=True)         # --- rank acording to the given col
```

```python
sample['New_col2'] = sample['Quantity'].cumsum()     # --- add things as we go down 
```
