## Removing unwanted data

```python
data = data.drop(['PassengerId' , 'Name' ,  'Ticket' , 'Cabin'] , axis=1)
```

## Filling null values

```python
x = data.iloc[: , 1:8].values
y = data.iloc[: , 0].values  
```

```python
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(missing_values=np.nan, strategy='mean')
imputer.fit(x[:, 2:3])
x[:, 2:3] = imputer.transform(x[:, 2:3])
```

```python
data
```

## Splitting of data

```python
from sklearn.model_selection import train_test_split
x_train , x_test, y_train , y_test = train_test_split(x,y,test_size=0.20,random_state=41)
```

```python
x_train
```

## Feature scaling

```python
from sklearn.preprocessing import StandardScaler
st = StandardScaler()
x_train[: , [2,5]]= st.fit_transform(x_train[: , [2,5]])
x_test[: , [2,5]] = st.transform(x_test[: , [2,5]])
```

## Encoding the data

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
ct = ColumnTransformer(transformers=[('encoder' , OneHotEncoder(handle_unknown='ignore') , [1,6])] , remainder='passthrough')
x_train = np.array(ct.fit_transform(x_train))
x_test = np.array(ct.transform(x_test))
```

## Training and fitting of data

```python
from sklearn.svm import SVC
model = SVC(kernel = 'rbf', random_state = 0)
model.fit(x_train, y_train)
```

## Model prediction

```python
y_pred  = model.predict(x_test)
```

```python
print(np.concatenate((y_pred.reshape(len(y_pred),1) , y_test.reshape(len(y_test), 1)) , 1))
```

## Accuracy and Confusion Matirx

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test , y_pred)
```

```python
from sklearn.metrics import confusion_matrix
cm  =confusion_matrix(y_test , y_pred)
print(cm)
```

```python
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(5,4))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Not Survived", "Survived"],
    yticklabels=["Not Survived", "Survived"]
)

plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.title("Confusion Matrix – Titanic Survival Prediction")
plt.show()
```
